```python
# method to get prediction for watermark
import torch
import sys
sys.path.append("./lm-watermarking") ## when add to any folder path to the code

from transformers import (AutoTokenizer,
                          AutoModelForSeq2SeqLM,
                          AutoModelForCausalLM,
                          LogitsProcessorList)

from watermark_processor import WatermarkDetector

model_name_or_path = "google-t5/t5-small"

tokenizer = AutoTokenizer.from_pretrained(model_name_or_path)
gamma = 0.25
seeding_scheme = "simple_1"
device =  "cuda" if torch.cuda.is_available() else "cpu"
detection_z_threshold = 4.0
normalizers = ""
ignore_repeated_bigrams = False
select_green_tokens = True
delta = 2.0


watermark_detector = WatermarkDetector(
        vocab=list(tokenizer.get_vocab().values()),
        gamma=gamma,
        seeding_scheme=seeding_scheme,
        device=device,
        tokenizer=tokenizer,
        z_threshold=detection_z_threshold,
        normalizers=normalizers,
        ignore_repeated_bigrams=ignore_repeated_bigrams,
        select_green_tokens=select_green_tokens)

def get_wm_pred(article):
    # purpose is to send article to wm detector and return prediction either 0/1, here for human generated texts - 0
    score_dict = watermark_detector.detect(article)
    if score_dict['prediction']:
        return 1
    else:
        return 0

# article = train_ds[0]['article']

# get_wm_pred(article)

wm_pred_list = []
take_sample = 20

for i in range(take_sample):
    article = train_ds[i]['article']
    if len(article)-1 > watermark_detector.min_prefix_len:
        wm_pred = get_wm_pred(article)
        wm_pred_list.append(wm_pred)
    else:
        print("Error: string too short to compute metrics")


wm_pred_list
```

```python
# generate summarized texts.

# Load model directly
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
from watermark_processor import WatermarkLogitsProcessor

# tokenizer = AutoTokenizer.from_pretrained("google-t5/t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google-t5/t5-small")


watermark_processor = WatermarkLogitsProcessor(
    vocab=list(tokenizer.get_vocab().values()),
    gamma=gamma,
    delta=delta,
    seeding_scheme=seeding_scheme,
    select_green_tokens=select_green_tokens)


def get_summarize_text(text, max_length=200, **kwargs):
    # Prepend the "summarize:" prefix
    input_text = "summarize: " + text

    # Encode the input text
    encoded_input = tokenizer.encode(input_text, return_tensors="pt")

    # Generate the summary
    summary_tokens = model.generate(
        encoded_input,
        logits_processor=LogitsProcessorList([watermark_processor]),
        max_length=max_length,
        **kwargs)

    # Decode the summary
    summary_text = tokenizer.decode(
        summary_tokens[0], skip_special_tokens=True)

    return summary_text


article = train_ds[4]['article']

get_summarize_text(article, min_length=100)
```

```python
### generate human + summarize text prediction

take_sample = 200

human_y_true = []
human_y_pred = []

ai_y_true = []
ai_y_pred = []


for i in range(take_sample):

    article = train_ds[i]['article']

    # human prediction
    if len(article)-1 > watermark_detector.min_prefix_len:
        human_wm_pred = get_wm_pred(article)

        human_y_true.append(0)
        human_y_pred.append(human_wm_pred)

        print(0, human_wm_pred)
    else:
        print("Error: string too short to compute metrics")


    # AI prediction
    summarize_text = get_summarize_text(article, max_length= 300, num_beams=1)
    if len(summarize_text)-1 > watermark_detector.min_prefix_len:
        ai_wm_pred = get_wm_pred(summarize_text)

        ai_y_true.append(1)
        ai_y_pred.append(ai_wm_pred)

        print(1, ai_wm_pred)
    else:
        print("Error: string too short to compute metrics")

    print(f'Processed sample: {i}/{take_sample}')

# print(human_y_true)
# print(human_y_pred)

# print(ai_y_true)
# print(ai_y_pred)
```

```python
## evaluate confusion matrix and classification report

from sklearn.metrics import confusion_matrix
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

y_true = human_y_true + ai_y_true
y_pred = human_y_pred + ai_y_pred


cm = confusion_matrix(y_true, y_pred)


# Display the confusion matrix
plot_cm = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["Human", "Watermarked"])
plot_cm.plot()
plt.show()
```

```python
## classification report/matrix with precision, recall, f1-score

from sklearn.metrics import classification_report

y_true = y_true
y_pred = y_pred
target_names = ['Human Text (Non-Watermarked)', 'AI Text (Watermarked)']
report_cnn = classification_report(y_true, y_pred, target_names=target_names)

print(report_cnn)
```
